### Ejercicio 1 — Tipos de datos y operaciones básicas
Pide al usuario su nombre y edad. Muestra: 'Hola, Ana. Dentro de 10 años tendrás 25 años.'

```python
def saludo(nombre: str, edad: int) -> str:
    edad_nueva = edad + 10

    return f"Hola, {nombre}. Dentro de 10 años tendrás {edad_nueva} años."

```

```python
assert saludo("Ana", 15) == "Hola, Ana. Dentro de 10 años tendrás 25 años."
assert saludo("Luis", 20) == "Hola, Luis. Dentro de 10 años tendrás 30 años."
print("✅ Ejercicio 1 correcto")
```

### Ejercicio 2 — Cálculo con operadores
Pide un número entero y muestra su doble, su triple y su cuadrado.

```python
def operaciones_basicas(n: int) -> tuple:
    doble = n * 2
    triple = n * 3
    cuadrado = n ** 2
    return (doble, triple, cuadrado)
```

```python
assert operaciones_basicas(2) == (4, 6, 4)
assert operaciones_basicas(5) == (10, 15, 25)
print("✅ Ejercicio 2 correcto")
```

### Ejercicio 3 — Letra del DNI
Dado un número de DNI (sin letra), calcula su letra correspondiente.

```python
def letra_dni(numero: int) -> str:
    # Calcula la letra del DNI usando el módulo 23
    letras = "TRWAGMYFPDXBNJZSQVHLCKE"
    resto = numero % 23
    return letras[resto]


assert letra_dni(12345678) == 'Z'
assert letra_dni(11111111) == 'H'

print("✅ Ejercicio 3 correcto")
```

```python
assert letra_dni(12345678) == 'Z'
assert letra_dni(11111111) == 'H'
print("✅ Ejercicio 3 correcto")
```

### Ejercicio 5 — Match (estructuras de selección múltiples)
Pide un número del 1 al 7 y muestra el nombre del día de la semana.

```python
def dia_semana(n: int) -> str:
    match n:
        case 1:
            return "Lunes"
        case 2:
            return "Martes"
        case 3:
            return "Miércoles"
        case 4:
            return "Jueves"
        case 5:
            return "Viernes"
        case 6:
            return "Sábado"
        case 7:
            return "Domingo"
        case _:
            return "Día no válido"
```

```python
assert dia_semana(1) == 'Lunes'
assert dia_semana(7) == 'Domingo'
print('✅ Ejercicio 5 correcto')
```

### Ejercicio 6 — Rango y suma de números pares
Suma todos los números pares entre 1 y 50.

```python
def suma_pares() -> int:
    total = 0
    for i in range(1, 51):
        if i % 2 == 0:
            total += i
    return total
```

```python
assert suma_pares() == sum(i for i in range(1,51) if i%2==0)
print('✅ Ejercicio 6 correcto')
```

### Ejercicio 7 — Contar vocales
Pide una palabra y devuelve cuántas vocales contiene.

```python
def contar_vocales(palabra: str) -> int:
    contador = 0
    for letra in palabra.lower():
        if letra in "aeiouáéíóú":
            contador += 1
    return contador
```

```python
assert contar_vocales('hola') == 2
assert contar_vocales('murciélago') == 5
print('✅ Ejercicio 7 correcto')
```

### Ejercicio 8 — Uso de break y continue
Pide una lista de números y devuelve la suma hasta encontrar el número 0 (sin incluirlo).

```python
def suma_hasta_cero(numeros: list[int]) -> int:
    total = 0
    for n in numeros:
        if n == 0:
            break
        total += n
    return total
```

```python
assert suma_hasta_cero([1,2,3,0,5]) == 6
assert suma_hasta_cero([5,5,0,10]) == 10
print('✅ Ejercicio 8 correcto')
```

### Ejercicio 9 — Excepciones
Pide dos números y devuelve su división. Si el divisor es 0, devuelve el texto 'Error: división por cero'.

```python
def dividir(a: float, b: float) -> float | str:
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: división por cero"
```

```python
assert dividir(10,2) == 5
assert dividir(8,0) == 'Error: división por cero'
print('✅ Ejercicio 9 correcto')
```

### Ejercicio 10 — Juego de adivinanza
Genera un número secreto del 1 al 10. El jugador intenta adivinarlo. Devuelve 'Correcto' si acierta y 'Incorrecto' si no.

```python
def adivinar(num: int, secreto: int) -> str:
    if num == secreto:
        return "Correcto"
    else:
        return "Incorrecto"
```

```python
assert adivinar(5,5) == 'Correcto'
assert adivinar(3,7) == 'Incorrecto'
print('✅ Ejercicio 10 correcto')
```
